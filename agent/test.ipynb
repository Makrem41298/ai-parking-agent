```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from agent.vectorstore.vectorstore import VectorStore
from agent.graph.graph_builder import GraphBuilder


load_dotenv()

GROQ_API_KEY = os.getenv("GROQ_API_KEY")

if not GROQ_API_KEY:
    raise ValueError("GROQ_API_KEY is missing in .env file")

os.environ["GROQ_API_KEY"] = GROQ_API_KEY


model = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0,
    reasoning_format="parsed",
    max_retries=2,
)


```

```python
import os

print(os.path.abspath("./agent/storage/chunks.pkl"))
print(os.path.exists("storage/chroma_db"))
```

```python

vector_store = VectorStore(chunks_path="storage/chunks.pkl",persist_directory="storage/chroma_db")
vector_store.setup()

graph_builder = GraphBuilder(
    llm=model,
    vector_store=vector_store
)
```

```python
from schemas.agent_schema import ModeResponse

result = graph_builder.run_graph(
        question="when i can refoun money",
        user_id=None,
        roleUser=None,
        reclamation_id=None,
        session_id=None,
        mode_response=ModeResponse.reclamation_response,
        number_vectors=vector_store.count_vectors()

    )
result.get("message", "")
```
